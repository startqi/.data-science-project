```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
df = pd.read_csv('movie_scores.csv')
df.head()
```

```python
fig, ax = plt.subplots(figsize=(10, 6), dpi=100)

x = np.arange(len(df['MovieTitle']))
width = 0.35

ax.bar(x - width/2, df['Tomatometer'], width, label='Tomatometer')
ax.bar(x + width/2, df['AudienceScore'], width, label='AudienceScore')

ax.set_xticks(x)
ax.set_xticklabels(df['MovieTitle'], rotation=15)

ax.set_yticks(np.arange(0, 101, 20))
ax.set_yticks(np.arange(0, 101, 5), minor=True)

ax.yaxis.grid(which='major', linestyle='-')
ax.yaxis.grid(which='minor', linestyle='--')

ax.set_xlabel('Movie')
ax.set_ylabel('Score (%)')
ax.set_title('Movie Comparison: Tomatometer vs Audience Score')
ax.legend()

plt.show()
```
