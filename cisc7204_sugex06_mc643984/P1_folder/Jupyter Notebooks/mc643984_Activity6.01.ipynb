```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

%matplotlib inline

googl = pd.read_csv('GOOGL_data.csv')
fb = pd.read_csv('FB_data.csv')
aapl = pd.read_csv('AAPL_data.csv')
amzn = pd.read_csv('AMZN_data.csv')
msft = pd.read_csv('MSFT_data.csv')

for df in [googl, fb, aapl, amzn, msft]:
    df['date'] = pd.to_datetime(df['date'])

plt.figure(figsize=(14, 7), dpi=100)

plt.plot(googl['date'], googl['close'], label='Google')
plt.plot(fb['date'], fb['close'], label='Facebook')
plt.plot(aapl['date'], aapl['close'], label='Apple')
plt.plot(amzn['date'], amzn['close'], label='Amazon')
plt.plot(msft['date'], msft['close'], label='Microsoft')

plt.xlabel('Date')
plt.ylabel('Closing Price (USD)')
plt.title('Stock Trends: Big Five Companies (Past 5 Years)')
plt.legend()
plt.grid()

ax = plt.gca()
ax.xaxis.set_major_locator(mdates.MonthLocator(interval=3))   
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m'))   

plt.xticks(rotation=45)    
plt.tight_layout()        
plt.show()
```
