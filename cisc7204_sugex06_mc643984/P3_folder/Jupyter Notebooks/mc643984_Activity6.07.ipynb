```python
import os
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
%matplotlib inline
```

```python
image_files = ['photo-1.jpg', 'photo-2.jpg', 'photo-3.jpg', 'photo-4.jpg']

labels = ['coast', 'beach', 'building', 'city at night']
```

```python
images = [mpimg.imread(f) for f in image_files]

# 创建 2x2 网格
fig, axes = plt.subplots(2, 2, figsize=(10, 10), dpi=100)

for i, ax in enumerate(axes.flat):
    ax.imshow(images[i])
    ax.set_xlabel(labels[i], fontsize=14)  
    ax.set_xticks([])                 
    ax.set_yticks([])                       
    ax.set_frame_on(False)               

plt.tight_layout()
plt.show()
```
