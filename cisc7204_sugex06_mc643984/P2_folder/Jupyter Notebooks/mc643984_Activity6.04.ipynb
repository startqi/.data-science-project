```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
data = pd.read_csv('smartphone_sales.csv')
data.head()
```

```python
plt.figure(figsize=(12, 6), dpi=100)

data.set_index('Quarter').plot.area(ax=plt.gca(), alpha=0.7)

plt.xlabel('Quarter')
plt.ylabel('Sales Units (Millions)')
plt.title('Smartphone Sales Comparison by Quarter')
plt.legend(loc='upper left')
plt.tight_layout()
plt.show()
```
