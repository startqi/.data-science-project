```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv'
bills = pd.read_csv(url)
bills.head()
```

```python
matrix = bills.pivot_table(index='day', columns='smoker', values='total_bill', aggfunc='sum')
matrix
```

```python
plt.figure(figsize=(10, 6), dpi=100)

matrix.plot(kind='bar', stacked=True, ax=plt.gca())

plt.xlabel('Day')
plt.ylabel('Total Bills (USD)')
plt.title('Restaurant Performance by Day (Smoker vs Non-Smoker)')
plt.legend(title='Smoker')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
